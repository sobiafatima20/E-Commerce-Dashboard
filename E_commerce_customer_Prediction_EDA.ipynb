###Outliers

```python
numerical_data = commerce.select_dtypes(include=['number'])

numerical_data.head()
```

```python
plt.figure(figsize=(15,8))

for i, column in enumerate(numerical_data.columns, 1):
  plt.subplot(2, 3, i)
  sns.boxplot(y=numerical_data[column])
  plt.title(f'Box Plot of {column}')
  plt.ylabel(column)

plt.tight_layout()
plt.show()
```

Box Plot Analysis

Box plots were created for all five numerical columns to identify potential outliers and understand data distribution.

Avg. Session Length: A few potential outliers are visible on the upper side.

Time on App: Potential outliers are present on both the upper and lower sides.

Time on Website: A few potential outliers can be seen on both sides.

Length of Membership: Several potential outliers are visible on both the upper and lower sides.

Yearly Amount Spent: Potential outliers are present at both lower and upper spending values.

Conclusion: The box plots show that some numerical columns contain potential outliers. These values will be investigated using the IQR method before deciding whether to keep or remove them.

```python
plt.figure(figsize=(15, 10))

for i, column in enumerate(numerical_data.columns, 1):
  plt.subplot(2, 3, i)
  sns.histplot(data=commerce, x=column, kde=True)
  plt.title(f'Histogram of {column}')
  plt.xlabel(column)
  plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

Histograms were created for all five numerical columns to understand their data distributions.

Avg. Session Length: The distribution is approximately bell-shaped and fairly symmetric.

Time on App: Most values are concentrated around 11–13, with an approximately bell-shaped distribution.

Time on Website: Most values are concentrated around 36–38 and show a fairly symmetric distribution.

Length of Membership: Most customers have membership values around 3–4, with a roughly bell-shaped distribution.

Yearly Amount Spent: Most customers spend between approximately 400 and 600. The distribution has a slight right tail.

```python

q1 = commerce["Length of Membership"].quantile(0.25)
q3 = commerce["Length of Membership"].quantile(0.75)

upper = q3 + 1.5 * (q3 - q1)
lower = q1 - 1.5 * (q3 - q1)

upper_outliers = commerce[commerce["Length of Membership"] > upper]
lower_outliers = commerce[commerce["Length of Membership"] < lower]

print("Upper limit:", upper, "| Upper Outliers:", len(upper_outliers))
print("Lower limit:", lower, "| Lower Outliers:", len(lower_outliers))
print("Total Outliers:", len(upper_outliers) + len(lower_outliers))
```

```python

q1 = commerce["Yearly Amount Spent"].quantile(0.25)
q3 = commerce["Yearly Amount Spent"].quantile(0.75)

upper = q3 + 1.5 * (q3 - q1)
lower = q1 - 1.5 * (q3 - q1)

upper_outliers = commerce[commerce["Yearly Amount Spent"] > upper]
lower_outliers = commerce[commerce["Yearly Amount Spent"] < lower]

print("Upper limit:", upper, "| Upper Outliers:", len(upper_outliers))
print("Lower limit:", lower, "| Lower Outliers:", len(lower_outliers))
print("Total Outliers:", len(upper_outliers) + len(lower_outliers))
```

###IQR Method for Outlier Detection

The Interquartile Range (IQR) method was applied to identify potential outliers in two numerical columns.

Length of Membership:

Lower Limit: 1.1364

Upper Limit: 5.9206

Number of Outliers: 12

Yearly Amount Spent:

Lower Limit: 288.6250

Upper Limit: 705.7272

Number of Outliers: 9

The IQR method identified 12 potential outliers in Length of Membership and 9 potential outliers in Yearly Amount Spent.

###Decision About Outliers

The identified outliers will be kept in the dataset because they may represent genuine customer behavior. Removing them could result in losing valuable information about customers with unusually high or low membership durations and spending patterns.

Q7. Which platform, App or Website, has a stronger relationship with Yearly Amount Spent?

```python
platform_correlation = pd.Series({
    'App': commerce['Time on App'].corr(
        commerce['Yearly Amount Spent']
    ),
    'Website': commerce['Time on Website'].corr(
        commerce['Yearly Amount Spent']
    )
})

print(platform_correlation)

platform_correlation.plot(
    kind='bar',
    figsize=(8, 5),
    title='App vs Website Correlation with Yearly Spending'
)

plt.xlabel('Platform')
plt.ylabel('Correlation with Yearly Amount Spent')
plt.xticks(rotation=0)
plt.show()
```

Time on App has a correlation of approximately 0.499 with yearly spending, while Time on Website has a correlation of approximately -0.003. Therefore, the App has a stronger positive relationship with yearly spending.

Q8. Who are the top 10 customers with the highest Yearly Amount Spent?

```python
top_10_customers = commerce.nlargest(
    10,
    'Yearly Amount Spent'
)[['Email', 'Yearly Amount Spent']]

print(top_10_customers)
```

The table identifies the ten customers with the highest yearly spending. The highest recorded yearly spending is approximately 765.52.
